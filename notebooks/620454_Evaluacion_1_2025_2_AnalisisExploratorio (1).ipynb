# Análisis de simetría/asimetría

## Variable elegida por el equipo que será analizada

`````python
data.hour_of_day.describe().round(2)
`````

`````python
moda = data.groupby('hour_of_day')['hour_of_day'].count().sort_values(ascending = False).index[0]
mediana = np.quantile(data.hour_of_day, 0.5)
media = np.mean(data.hour_of_day)
print("Moda {0} - Mediana {1} - Media {2:.1f}".format(moda, mediana, media))
`````

`````python
sns.histplot(data.hour_of_day, color='blue', edgecolor='black', kde=True)
plt.title('Distribución de la hora de compra', fontsize=18, fontweight="bold")
plt.xlabel('Hora de compra', fontsize=16, fontweight="bold")
plt.ylabel('Cantidad de compras de café', fontsize=16, fontweight="bold")
plt.show()
`````

---
✅ **Interpretación**

---


+ El gráfico de las ventas de café describe la siguiente distribución de frecuencia:
  + Se observa un punto máximo a las 10 h, que concentra el mayor índice de consumo en la mañana. Sin embargo, la mediana se ubica a las 14 h y la media en 14.2 h, lo que inclina el gráfico hacia el mediodía y las primeras horas de la tarde. Esto indica que el consumo de café no es exclusivo de la mañana.

    La cercanía entre la media y la mediana sugiere una distribución de frecuencias casi simétrica, aunque con una ligera asimetría positiva, causada por una acumulación de compras en la tarde.
